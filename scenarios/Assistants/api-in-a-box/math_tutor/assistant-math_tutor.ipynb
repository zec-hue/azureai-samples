### Parameters

```python
import os

from dotenv import load_dotenv

load_dotenv("../.env")  # make sure to have the .env file in the root directory of the project
api_endpoint = os.getenv("OPENAI_URI")
api_key = os.getenv("OPENAI_KEY")
api_version = os.getenv("OPENAI_VERSION")
api_deployment_name = os.getenv("OPENAI_GPT_DEPLOYMENT")

should_cleanup: bool = True
```

## Run this Example

## Load the required libraries

```python
import sys
from pathlib import Path

sys.path.append(str(Path().resolve().parent / 'common'))

from openai import AzureOpenAI
from common import Common
```

### Create an Azure OpenAI client

```python
client = AzureOpenAI(api_key=api_key, api_version=api_version, azure_endpoint=api_endpoint)
```

### Create an Assistant and a Thread

```python
assistant = client.beta.assistants.create(
    name="Math Tutor",
    instructions="You are a personal math tutor. Write and run code to answer math questions.",
    tools=[{"type": "code_interpreter"}],
    model=api_deployment_name,
)

thread = client.beta.threads.create()
```

### Format and display the Assistant Messages for text and images

### Process the user messages

```python
common = Common(client, assistant, thread)
```

### Have a conversation with the Assistant

```python
common.process_message("What is the linear equation when two (x,y) points are (1,1) and (5,10)?")
```

```python
common.process_message("I need to solve the equation `3x + 11 = 14`. Can you help me?")
```

```python
common.process_message(
    """x=r*cos(u)sin(v)
y=r*sin(u)sin(v)
r=2+sin(7*u+5*v) 

for 0<u<2π and 0<v<π.

Create a graph of the equation z=r*cos(v)."""
)
```

## Cleaning up

```python
if should_cleanup:
    client.beta.assistants.delete(assistant.id)
    client.beta.threads.delete(thread.id)
```
